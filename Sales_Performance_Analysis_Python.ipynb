```python
df = pd.read_excel("Documents/Sales_Performance_Analysis_Project/Google_Sheets_Analysis/Sales_data (1).xlsx")

print("Sales dataset loaded successfully!")
print("Dataset size:", df.shape)

df.head()
```

```python
print("Dataset information:")
df.info()

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())
```

```python
print("MISSING VALUES")
print(df.isnull().sum())

print("\nTOTAL MISSING VALUES:")
print(df.isnull().sum().sum())

print("\nDUPLICATE ROWS:")
print(df.duplicated().sum())

print("\nLAST 10 ROWS:")
display(df.tail(10))
```

```python
# Keep the original dataset unchanged
df_clean = df.copy()

# Remove completely empty rows
df_clean = df_clean.dropna(how="all")

# Remove incomplete spreadsheet artefacts:
# rows with no order date and no other business data
business_columns = [
    col for col in df_clean.columns if col != "Order_ID"
]

df_clean = df_clean.dropna(
    subset=business_columns,
    how="all"
)

# Remove exact duplicate records
df_clean = df_clean.drop_duplicates()

# Reset row numbering
df_clean = df_clean.reset_index(drop=True)

print("CLEANING COMPLETED")
print("Clean dataset shape:", df_clean.shape)
print("\nRemaining missing values:")
print(df_clean.isnull().sum())
print("\nRemaining duplicates:", df_clean.duplicated().sum())
```

```python
# Key Performance Indicators (KPIs)

total_revenue = df_clean["Revenue_NGN"].sum()
total_units = df_clean["Quantity"].sum()
average_revenue = df_clean["Revenue_NGN"].mean()
highest_sale = df_clean["Revenue_NGN"].max()
lowest_sale = df_clean["Revenue_NGN"].min()
total_records = len(df_clean)

print("SALES PERFORMANCE KPIs")
print("-" * 35)
print(f"Total Revenue: ₦{total_revenue:,.2f}")
print(f"Total Units Sold: {total_units:,.0f}")
print(f"Average Revenue per Record: ₦{average_revenue:,.2f}")
print(f"Highest Sale: ₦{highest_sale:,.2f}")
print(f"Lowest Sale: ₦{lowest_sale:,.2f}")
print(f"Total Sales Records: {total_records}")
```

```python
category_analysis = (
    df_clean.groupby("Product_Category")
    .agg(
        Total_Revenue=("Revenue_NGN", "sum"),
        Units_Sold=("Quantity", "sum"),
        Number_of_Records=("Order_ID", "count")
    )
    .sort_values("Total_Revenue", ascending=False)
)

display(category_analysis)
```

```python
# Monthly revenue analysis
df_clean["Month"] = df_clean["Order_Date"].dt.to_period("M")

monthly_sales = (
    df_clean.groupby("Month")["Revenue_NGN"]
    .sum()
    .reset_index()
)

monthly_sales["Month"] = monthly_sales["Month"].astype(str)

display(monthly_sales)

# Create professional chart
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=monthly_sales,
    x="Month",
    y="Revenue_NGN",
    marker="o",
    linewidth=2.5
)

plt.title("Monthly Sales Revenue Trend - 2025", fontsize=15)
plt.xlabel("Month")
plt.ylabel("Revenue (NGN)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import os

output_folder = "Documents/Sales_Performance_Analysis_Project/Python_Analysis"
os.makedirs(output_folder, exist_ok=True)

# 1. Monthly revenue
plt.figure(figsize=(10, 5))
sns.lineplot(data=monthly_sales, x="Month", y="Revenue_NGN", marker="o")
plt.title("Monthly Sales Revenue Trend")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig(os.path.join(output_folder, "Monthly_Revenue_Trend.png"))
plt.show()

# 2. Category revenue
plt.figure(figsize=(8, 5))
sns.barplot(data=category_analysis.reset_index(),
            x="Product_Category", y="Total_Revenue")
plt.title("Revenue by Product Category")
plt.tight_layout()
plt.savefig(os.path.join(output_folder, "Category_Revenue.png"))
plt.show()

# 3. Regional revenue
region_sales = df_clean.groupby("Region", dropna=False)["Revenue_NGN"].sum().sort_values(ascending=False)
plt.figure(figsize=(9, 5))
region_sales.plot(kind="bar")
plt.title("Revenue by Region")
plt.ylabel("Revenue (NGN)")
plt.tight_layout()
plt.savefig(os.path.join(output_folder, "Regional_Revenue.png"))
plt.show()

# 4. Sales representative revenue
rep_sales = df_clean.groupby("Sales_Rep", dropna=False)["Revenue_NGN"].sum().sort_values(ascending=False)
plt.figure(figsize=(9, 5))
rep_sales.plot(kind="bar")
plt.title("Revenue by Sales Representative")
plt.ylabel("Revenue (NGN)")
plt.tight_layout()
plt.savefig(os.path.join(output_folder, "Sales_Rep_Performance.png"))
plt.show()

# Save cleaned dataset
df_clean.to_csv(os.path.join(output_folder, "Cleaned_Sales_Data.csv"), index=False)

print("Charts and cleaned dataset saved successfully!")
```

```python
# Top 5 products by revenue
product_sales = (
    df_clean.groupby("Product")["Revenue_NGN"]
    .sum()
    .sort_values(ascending=False)
)

print("TOP 5 PRODUCTS BY REVENUE")
display(product_sales.head(5))

# Regional performance
print("\nREVENUE BY REGION")
display(
    df_clean.groupby("Region")["Revenue_NGN"]
    .sum()
    .sort_values(ascending=False)
)

# Sales representative performance
print("\nREVENUE BY SALES REPRESENTATIVE")
display(
    df_clean.groupby("Sales_Rep")["Revenue_NGN"]
    .sum()
    .sort_values(ascending=False)
)

# Best-performing month
print("\nBEST SALES MONTH")
display(monthly_sales.loc[monthly_sales["Revenue_NGN"].idxmax()])
```
